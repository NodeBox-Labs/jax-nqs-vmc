# Quickstart: Neural Quantum States in 5 minutes

This notebook trains a neural network to find the ground state of the 1D Harmonic Oscillator, and compares it to the exact analytic solution.

```python
import jax
from nqs_vmc.config import RunConfig, AnsatzConfig, SamplerConfig, OptimizerConfig
from nqs_vmc.systems import build_system
from nqs_vmc.ansatz import build_ansatz
from nqs_vmc.trainer import train
import jax.numpy as jnp

cfg = RunConfig(
    system="harmonic",
    ansatz=AnsatzConfig(type="even", hidden=(32, 32)),
    sampler=SamplerConfig(n_chains=64, n_samples=4096, n_iter=300),
    optimizer=OptimizerConfig(name="adam", lr=0.01, n_iter=200),
)
system = build_system(cfg)
ansatz = build_ansatz(cfg.ansatz)

# Initialize
key = jax.random.PRNGKey(0)
params = ansatz.init(key, jnp.zeros((1, 1)))

# Train
state = train(lambda p, x: ansatz.apply(p, x), params, system, cfg, seed=0, verbose=True)
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(8, 4))
plt.plot(state.history["iteration"], state.history["energy"])
plt.axhline(y=0.5, color='red', linestyle='--', label="Exact E = 0.5")
plt.xlabel("Iteration")
plt.ylabel("Energy")
plt.title("VMC Convergence to HO Ground State")
plt.legend()
plt.show()
```
